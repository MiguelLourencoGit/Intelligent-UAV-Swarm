```python
import numpy as np

# Definir as posições iniciais dos UAVs 
# 120.0,500.0,
# 59.54855169804455,1.5227202548656282,20.561475412279616,42.77581139119352,0.3037399981834165,56.947888135775045,44.34573782095048,36.33570249394532,
# 90.6,48.4,90.6,48.4,90.6,48.4,90.0,48.4,
# 88.15507698693936,36.78870171016068,108.53147464836871,29.52509422397264,100.86584716454276,25.65262279576093,99.31958138916532,25.80815568941728



# # 59.54855169804455,1.5227202548656282,20.561475412279616,42.77581139119352,0.3037399981834165,56.947888135775045,44.34573782095048,36.33570249394532,
initial_positions = np.array([
    [59.54855169804455,1.5227202548656282],
    [20.561475412279616,42.77581139119352],
    [0.3037399981834165,56.947888135775045],
    [44.34573782095048,36.33570249394532]
])


# preenche com 90.6,48.4,90.6,48.4,90.6,48.4,90.0,48.4,

final_positions = np.array([
    [90.6,48.4],    
    [90.6,48.4],
    [90.6,48.4],
    [90.0,48.4],
])

# Função para interpolar entre as posições
def interpolate_positions(initial_positions, final_positions, steps=6):
    all_positions = []
    
    # Para cada "tempo", gerar a posição de todos os UAVs
    for t in range(steps):
        interpolated_positions_at_t = []
        for i in range(initial_positions.shape[0]):
            start = initial_positions[i]
            end = final_positions[i]
            
            # Interpolação linear
            alpha = t / (steps - 1)
            interpolated_position = start + alpha * (end - start)
            
            # Adiciona a posição interpolada de cada UAV à lista para esse "tempo"
            interpolated_positions_at_t.extend(interpolated_position)
        
        # Adiciona as posições de todos os UAVs para esse tempo
        all_positions.append(interpolated_positions_at_t)
    
    return np.array(all_positions)

# Gerar as posições interpoladas
interpolated_positions = interpolate_positions(initial_positions, final_positions)

# Mostrar os resultados em uma única linha, como solicitado
# Flatten a matriz para exibir tudo em uma linha contínua
interpolated_positions_flat = interpolated_positions.flatten()
print(" ".join(map(str, interpolated_positions_flat)))
```

```python
import re
import matplotlib.pyplot as plt

# Lê o log que colaste
with open("rl1.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Extrai timesteps e rewards médios
timesteps = [int(x) for x in re.findall(r"total_timesteps\s*\|\s*([\d\.]+)", text)]
rewards = [float(x) for x in re.findall(r"ep_rew_mean\s*\|\s*([\d\.]+)", text)]

plt.plot(timesteps, rewards)
plt.xlabel("Total Timesteps")
plt.ylabel("Average Reward")
#plt.title("Evolução do Reward Médio durante o Treino (PPO)")
plt.grid(True)
plt.ylim(7, 16)
plt.show()
```
